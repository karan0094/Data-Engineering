```python
#csv
source_df_path="/Volumes/ln_spark/files_schema/differentfiles/sample.csv"
source_df=spark.read.format("csv").option("header","true").option("inferSchema","true").load(source_df_path)
# source_df.printSchema()
source_df.write.format("delta").option("mergeSchema","true").mode("overwrite").saveAsTable("ln_spark.files_schema.employes")
```

```python
%sql
select * from ln_spark.files_schema.employes;
```

```python
#avro
source_df_path="/Volumes/ln_spark/files_schema/differentfiles/sample.avro"
source_df=spark.read.format("avro").option("infer_schema","true").load(source_df_path)
source_df.printSchema()
source_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("ln_spark.files_schema.employes")
```

```python
%sql
select * from ln_spark.files_schema.employes;
```

```python
source_df_path="/Volumes/ln_spark/files_schema/differentfiles/sample.json"
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,LongType
schema=StructType(
   [ StructField("name",StringType(),True),
    StructField("age",IntegerType(),True),
    StructField("city",StringType(),True),
    StructField("salary",LongType(),True),
    StructField("department",StringType(),True)]
)
source_df=spark.read.format("json").option("multiLine","true").schema(schema).load(source_df_path)
source_df.printSchema()
source_df.write.format("delta").mode("append").saveAsTable("ln_spark.files_schema.employes")
```

```python
%sql
select * from ln_spark.files_schema.employes;
```
